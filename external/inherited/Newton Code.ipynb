```python
def newton(f, fprime, p0, atol, max_iter):
    root_found = False

    for i in range(max_iter):
        p1 = p0 - f(p0)/fprime(p0)
        if abs(p1 - p0) <= atol:
            root_found = True
            return p1
        else:
            p0 = p1

    if not root_found:
        raise Exception(f"Newton's method did not converge within {max_iter} iterations. Try different initial guesses or increase max_iter.")
```

```python
p =  newton(f=lambda x: x**3+2*x**2+5*x-2, fprime=lambda x:3*x**2+4*x+5, p0=1, atol=1e-3, max_iter=20)
print(f"The root found by Newton's method is {p}.")
```
