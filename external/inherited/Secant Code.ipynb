```python
def secant(f, p0, p1, atol, max_iter):
    for i in range(max_iter):        
        p2 = p1 - f(p1) * (p1 - p0) / (f(p1) - f(p0))
        if abs(p2 - p1) <= atol:
            return p2
        else:
            p0, p1 = p1, p2

    raise Exception(f"Secant method did not converge within {max_iter} iterations. Try different initial guesses or increase max_iter.")
```

```python
p = secant(f=lambda x: x**3 + 2*x**2 + 5*x - 2, p0=0, p1=1, atol=1e-3, max_iter=20)
print(f"The root found by the secant method is {p}.")
```
