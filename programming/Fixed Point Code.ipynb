```python
g = lambda x: (2 - x**3 - 2*x**2) / 5

x = np.linspace(0,0.5,100)
y = g(x)

px.line(x=x, y=y)
```

```python
p = fixed_point(g, p0=0.3, atol=1e-3, max_iter=20)
print(f"The root found by fixed-point iteration is {p}.")
```
