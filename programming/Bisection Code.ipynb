```python
def bisection(f, a, b, atol, max_iter):
    root_found = False
    
    if f(a)*f(b) <= 0:
        for i in range(max_iter):
            p = (a+b)/2
            if b-p <= atol:
                root_found = True
                return p
            elif f(b)*f(p) <= 0:
                a = p
            else:
                b = p
        if not root_found:
            raise Exception(f"Bisection did not converge within {max_iter} iterations. Try a greater max_iter.")
    else:
        raise Exception("Bisection is not applicable for this problem.")
```

```python
root =  bisection(f=lambda x: x**3+2*x**2+5*x-2, a=0, b=2, atol=1e-3, max_iter=20)
print(f"The root found by bisection method is {root}.")
```
